```python
# 1) Use list comprehension to create a new list with **only alphabetic strings** (no numbers in them)

mixed_data = ["apple3", "banana", "grape5", "orange2", "guava"]

# Answer: 

words_only = [words for words in mixed_data if words.isalpha()]
print(words_only)

# Explanation:
# word.isalpha() returns True only if ALL characters are letters (no digits/ symbols)
# Filters out strings containing numbers like "apple3", "orange2"
# Essential for text processing in NLP/ data cleaning tasks. 
```

```python
# 2) Use dictionary comprehension to create a new dict with **only quarters having sales > 20,000**

sales = {"Q1": 15000, "Q2": 23000, "Q3": 18000, "Q4": 27000}

# Answer: 

high_sales = {k:v for k, v in sales.items() if v > 20000}
print(high_sales)

# Explanation: 
# sales.items() gives key-value pairs (k,v)
# if v > 20000, it filters quarters with high sales
# Creates new dict with same keys but only qualifying values, perfect for KPI filtering!
```

##### We can get the same output in FOR loop also then why WHILE loop? 
FOR loop is more like a iterator in Python. Here, we the users know the number of iterations of the loop.   
Basically we use WHILE loop in those situations when the user don't know how many times these statements would be executed. See the below example.

```python
list1 = [1,2,0,5,6]   # Generally an empty list is considered FALSE.So, list1 here is TRUE.

while list1:          # Means - "Keep executing the loop as long as list1 contains at least one item."
    print("Hi Jenny")  # See here do we users know how many times this program will get executed? No right? 
                       # If we run this alone, it will be an infinite loop. 
    list1.pop()        # Removes last element each time, 
    # and when the list1 gets empty after the five iteration,it becomes false hence while loop stopped there. 
```

##### WHILE LOOP WITH ELSE BLOCK
ELSE block would be executed when the WHILE block gets FALSE.   
BUT, ELSE block would not get executed when the WHILE loop forcefully terminated using BREAK.   
Let's see the examples below:

```python
count = 1
while count <= 5:
    print(count)
    count +=1    # Here, the loop ends naturally. 
else:
    print("In Else block")  # So, the ELSE block got executed. 

# You might have a question that why 'ELSE" block also runs here after the sucessful execution of WHILE loop?
# Note that - The else belongs to the while loop, not to an if statement. Easy rule to remember: 
# while...else: The else block runs when the loop ends naturally, but not when the loop is terminated by break. 
```

```python
count = 7   # 7 is greater than 5, so the condition is false and only ELSE statement executed. 
while count <= 5:
    print(count)
    count +=1
else:
    print("In Else block")
```

```python
count = 1
while count <= 5:
    print(count)
    count +=1
    if count == 3:
        break
else:
    print("In Else block")
print("Out from loop") # This statement is out from loop, so this can be executed. n

# Note that ELSE block would not get executed because the WHILE loop has been forcefully terminated by BREAK
# Also 3 not gets printed. BREAK statement ignores the last element always. Like here count == 3. 3 gets ignored. 
```

Easy way to remember:

##### LOGICAL OPERATORS

```python
# Logical operators includes ‘AND’, ‘OR’, ‘NOT’. 
# Here, ‘And & OR’ are binary logical operators however ‘NOT’ is an Unary logical operator. 
# And -> returns TRUE only when both conditions are TRUE otherwise it returns FALSE. 
# Or -> returns TRUE if one of the conditions is TRUE. Otherwise, it returns FALSE. 
# •	Not -> returns TRUE if the conditional expression returns FALSE and vice versa. 

# Let's see an example for AND:
x = 10
y = 20
z = 30

if x > y and y < z: # Here second part of condition is True however first part is False, hence the statement doesn't got executed. 
    print("x is the largest number")
else:
    print("x is not the largest number")
```

```python
# Let's see an example for OR:

x1 = 30
y1 = 50
z1 = 10

if x1 > y1 or x1 > z1:  # Here second part of condition is True eventhough first part is False, hence the statement got executed.  
    print("x1 is atleast larger than one number")
else:
    print("x1 is the smallest number")
```

```python
# Let's see an example for NOT:

x = 10
y = 20
z = 30

if not(x > y or x < z):  # Here both the conditions are FALSE. 'NOT' operator reverse it as TRUE, hence the statement got executed. 
    print("x is not the largest number")   # Change the OR operator as AND to see the difference. 
else:
    print("x is not the smallest number")
```

```python
# Difference between '==' vs "is" (Equality vs Identity)
# is → checks object identity: "Are these two variables pointing to the exact same object?"
# == → checks value equality: "Do these two objects contain the same values?"

list_a = [1, 2, 3]
list_b = list_a
list_c = [1, 2, 3]

# 'is' (Identity)
def check_identity(list1, list2):
    return list1 is list2

print(check_identity(list_a, list_b)) 
print(check_identity(list_a, list_c))

# '==' (Equality)
def check_equality(list1, list2):
    return list1 == list2

print(check_equality(list_a, list_b)) 
print(check_equality(list_a, list_c)) 

```

##### C) Strings and Bytes Built-ins: 

```python
# String (str) → human-readable text; 
# Bytes (bytes) → raw data in a **fixed, immutable form**. Bytes is useful when Python needs to work with **raw data**.
# Bytearray (bytearray) → **raw data that you can modify**

# 1) Bytes: Creates immutable bytes. Bytes is used to store data as a sequence of bytes.

text = "Hello"

c = bytes(text, "utf-8")
c1 = bytes(text, "utf-16")
c2 = bytes(text, "utf-32")
print(c)
print(c1)
print(c2)

# Remember: bytes is immutable — you cannot change individual bytes. 
# If you do any element modification using index, it will throw an error -> 'bytes' object does not support item assignment
```

```python
# 2. bytearray() — Creates changeable bytes; bytearray is similar to bytes, but it is mutable.

c3 = bytearray([65, 66, 67])  # bytearray only takes integer values not 'str' object 
print(c3)
c3[1] = 68
print(c3)
```

```python
# 3) memoryview() — Views bytes without copying them. memoryview() allows you to access the underlying data of a bytes-like object.
# memoryview() does NOT work with a normal Python string (str). 
# It works with objects that support the buffer protocol, such as bytes, bytearray, some ther binary objects. 

data = bytearray(b"Hello")
view = memoryview(data)
print(view[0])   # Output returns 72 because H is represented by byte 72 in ASCII/UTF-8.

# You can also modify the original data through the memory view.
data = bytearray(b"Hello")
data[0] = 74  # If we change the first element's bytearray from 72 ('H') to 74, the character also changed ('J') when you viewed it. 
print(data)

```

##### D) Iteration Built-ins: 

```python
# iter, next, aiter, anext, enumerate, zip, reversed, sorted, filter, map, all, any, range, slice

# 1) iter() -> iter() converts an iterable like a list into an iterator.
numbers3 = [10,20,30]
my_iterator = iter(numbers3)

print(next(my_iterator))  # next() -> next() gets the next value from an iterator.
print(next(my_iterator))
print(next(my_iterator))
# print(next(my_iterator))  # It gives an error because there is no further iteration
```

```python
# 2) enumerate() -> enumerate() gives you both the index and value while looping.

for index, i in enumerate(range(2, 6)):
    print(index, i)

# You can also start from 1

names1 = ["Anu", "Baama", "Charlie"]
for index, name in enumerate(names1, start = 1):
    print(index, name)
```

```python
# 3) zip() -> zip() combines two or more iterables element by element.

name4 = ["Anu", "Radha", "Ravi"]
age2 = [23,19,20]

for name , age in zip(name4, age2):
    print(name, age)

# You can also create a dictionary:
my_dict = dict(zip(name4, age2))
print(my_dict)

# Very useful for data processing.
```

```python
# 4) reversed() -> reversed() gives the items in reverse order.
age2 = [23,19,20]

for age in reversed(age2):
    print(age)

# you can convert this in to a list. 
list2 = list(reversed(age2))
print(list2)

# Important: reversed() does not modify the original list. It copies. 
```

```python
# 5) sorted() -> sorted() sorts values from ascending to descending and returns a new list.
age2 = [23,19,20]
sorted1 = sorted(age2)
print(sorted1)

# If you need descending to ascending, use 'reverse= True' argument
sorted2 = sorted(age2, reverse=  True)
print(sorted2)

# 6) filter() -> filter() keeps only the values that satisfy a condition.

# Example: Get only the even numbers using filter function
numbers4 = [1,2,3,4,5,6,7,8,9]
even = filter(lambda x : x % 2 == 0, numbers4)
print(list(even))

# Example 2: Get all the numbers greater than 4
condition = filter(lambda x: x > 4, numbers4)
print(list(condition))
```

```python
# 7) map() -> map() applies a function to every item.

# Example: Square every number.
numbers4 = [1,2,3,4,5,6,7,8,9]
square1 = map(lambda x: x **2, numbers4)
print(list(square1))

# Another example: Get new prices by increased each value by 1.20
price = [10,20,30]
new_price = map(lambda x: x + 1.20, price)
print(list(new_price))

# Remember:  map() → transform every value -> filter() → select some values
```

```python
# 8) all() -> all() checks whether EVERY value is True.

numbers4 = [1,2,3,4,5,6,7,8,9]
even1 = all(number % 2 == 0 for number in numbers4)  # 'all' takes exactly one arguement
print(even1)                                         # all values should be TRUE

numbers5 = [2,4,6,8,10]
even2 = all(number % 2 == 0 for number in numbers5)
print(even2)

# 9) any() -> any() checks whether ATLEAST ONE value is True.
numbers5 = [1,3,5,8]
even1 = any(number % 2 == 0 for number in numbers4)  # 'any' takes exactly one arguement
print(even1)                                         # Atleast ANY of the one value must be True

numbers6 = [1,3,5,7]
odd = any(number % 2 == 0 for number in numbers6)
print(odd)
```

```python
# 10) range() -> range() generates a sequence of numbers, commonly used with loops.
# start, stop, step
for i in range(2,11,2):
    print(i)

# reverse:
for i in range(10, 0, -2):
    print(i)
```

```python
# 1) classmethod() -> A classmethod is a method that receives the class itself as its first argument, usually called cls.

class Student:   
    school = "ABC school"   # Attribute

    @classmethod
    def show_school(cls):   # cls refers to the Student class.
        print(cls.school)

Student.show_school()  # calls the method directly using the class.

# Easy memory trick
# classmethod → works with CLASS
# cls         → represents CLASS
```

```python
# 2) staticmethod() -> A static method is a method that doesn't need the class or object.
# It behaves almost like a normal function, but we keep it inside the class because it is related to that class.

class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10,20))

# Notice there is no self and no cls.
# Easy memory trick: staticmethod → doesn't need object or class
```

```python
# 3) property() -> property() allows us to treat a method like an attribute.
# This sounds confusing initially, so let's see an example.

# Without property

class Student:
    def get_name(self):
        return "Dhivya"

student = Student()
print(student.get_name())  # We need parentheses because get_name is a method.

# With Property

class Student:

    @property
    def get_name(self):
        return "Dhivya"

student = Student()
print(student.get_name)  # Here, We don't use (). The method is being accessed like an attribute.


# Remember this sentence:
# classmethod → I need the CLASS.
# staticmethod → I need NEITHER class nor object.
# property → I need the OBJECT, but I want the method to LOOK LIKE AN ATTRIBUTE.
```

LESSON 10: SHALLOW VS DEEP COPY OF LISTS

```python
# Just a '=' operator allocation. 
list_A = [1,2,3]
list_B = list_A  # Refering the same memory location. 

print(list_A)
print(list_B)

# When make changes
list_B[2] = 100
print(list_A)
print(list_B)   # Due to the same memory location, changes happened in both places.
```

```python
# Shallow copy
list_A = [1,2,3]
list_B = list_A.copy() # Referring the different memory location

print(list_A)
print(list_B)

# When you make changes in a single dimension (not nested) of a Shallow copy
list_B[1] = 100
print(list_B)  #Due to the different memory location,changes only made here, not in the list_A. This is for NOT NESTED
print(list_A)
```

```python
# Shallow copy of Nested
list_C = [[1,2,3,4], [5,6,7,8]]
list_D = list_C.copy()

print(list_C)
print(list_D)

# When you make changes:
list_D[1][2] = 100
print(list_C)
print(list_D) # Eventhough different memory location, it refers to the same object inside it. 
              # Due to this, changes made in both places. 
```

```python
# Deep copy: A deep copy creates a completely independent copy. We use Python's copy module:

import copy
original = [[10,20], [30,40]]
deep = copy.deepcopy(original)
print(original)
print(deep)

# When changes made: 
deep[1][1] = 64
print(original)
print(deep)
```

HOW TO USE TUPLE PACKING AND UNPACKING? 


```python
emp = ("CB", "1992", "12", "27", 10000)

# You can also omit the parentheses:
emp = "CB", "1992", "12", "27", 10000  # Python still creates a tuple because commas separate the values.

# Tuple unpacking means assigning each tuple element to a separate variable.
location, year, month, date, salary  = emp
print(year)
print(salary)

# Extended unpacking
location, *_, salary = emp
print(location)
print(salary)
print(_)  # This takes all of the inbetween elements in a list. 
```

SETS:  

- Sets are unordered. Duplicates and indexing are not allowed. It may contain various data types. Whereas in Tuples and Lists, duplicates are allowed and these are ordered and indexing. 		                
- Why sets are not having indexing (sets are not sub-scriptable)? Because they do not have pre-defined ordered right! 		                  
- We can’t modify elements with indexing method; however, we can add, len, clear, pop, discard & remove items, membership operations in the sets. 

```python
set1 = {10, True, 'Jenny' , 1.11, 1, False, 0}
print(set1)  # Note in output that either 1 or True will be present according to computer language. 
             # Similarly, 0 or False. 
print(type(set1))

set2 = {}
print(type(set2))   # Note the type is dictionary not set. If you want to create an empty set. Do like below

set3 = set()
print(type(set3))
```

```python
# Add elements to a set
set4 = {10,23,26,54,10}
set4.add(99)
print(set4)

# Removing element in the set
set4.remove(10)
print(set4)

# set4.remove(87)  Throws 'KeyError: 87' because 87 did not present in the set. 
print(set4)

# Discarding element in the set. 
# Discard is similar to remove however if an element is not present in the set, it won't throw an error. 
set4.discard(87)  
print(set4)   # It simply returns the other elements in the set as output. 
```

```python
# Clearing elements in the set. 
set1.clear()
print(set1)

# Pop elements in the set : It will remove any random element in the set. 
print(set4.pop()) # It prints the item that's going to remove
print(set4)
```

```python
# We can add a tuple as an element but we can't add a list because list is mutable.
# Being tuple is immutable, and also due to sets can't do any modifications - tuples are allowed as an element inside it. 

set4.add((1,2,3))
print(set4)

# set4.add([6,7,8])
print(set4)   # TypeError: unhashable type: 'list'
```

Given a list of user IDs with duplicates, which Python data structure should you convert the list into to automatically remove duplicates and test for membership in O(1) average time complexity?

The correct answer is set in Python.  
A set automatically removes duplicate values and supports membership testing in O(1) average time complexity.  

Interview tip:  
Converting a list of \(n\) elements into a set takes O(n) average time, while checking membership in the resulting set takes O(1) average time.

WHAT ARE SET OPERATIONS (UNION, INTERSECTION, DIFFERENCE)

```python
set5 = {1,2,3}
set6 = {3,4,5}

# Union
print(set5.union(set6)) 
print(set5 | set6)   # same thing done in 'or' that is '|' (pipe symbol) operator. 

# Intersection
print(set5.intersection(set6))
print(set5 & set6)  # same thing done in 'and' that is '&' (and symbol) operator. 

# Difference
print(set5.difference(set6))  # Only returns the remaining elements in set5
print(set5 - set6)

print(set6.difference(set5))  # Only returns the remaining elements in set6
print(set6 - set5)  # same thing done in 'minus' that is '-' (minus symbol) operator. 
```

```python
# Symmetric Difference (XOR operation)  
# Means have to return all the elements from both the sets but NOT THE COMMON ELEMENT. 
# It's just an opposite of the intersection. Union - Intersection is the symmetric difference. 

set5 = {1,2,3}
set6 = {3,4,5}

print(set5.symmetric_difference(set6))
print(set6.symmetric_difference(set5))   
# Both returns the same result only. 

print(set5 ^ set6)  # same thing done in '^' that is 'XOR' operator. 
```

```python
# In Symmetric difference, we also have the option called 'update'
# In which we can update the difference included in the first set also. 

set5 = {1,2,3}
set6 = {3,4,5}

print(set5.symmetric_difference(set6))
print(set5)
print(set5.symmetric_difference_update(set6))  # This returns None
print(set5)                                    # Now set5 is updated with 1,2,4 & 5 but ignored the common element 3. 

set5 ^= set6  # XOR and equalto for Symmetric difference update. {1,2,4,5} ^= {3,4,5}
print(set5)
```

```python
#isdisjoint

set5 = {1,2,3}
set6 = {3,4,5}
print(set5.isdisjoint(set6)) # Disjoint is False because one common element is present between both the sets. 

set6 = {3,4,5}
set7 = {7,8,9}
print(set6.isdisjoint(set7)) # Disjoint is True because no common elements are there between both the sets. 
print(set6 & set7)  # returns empty set because there is no intersection right! 
```

```python
# issubset

set7 = {7,8,9}
set8 = {7,8,9,10}

print(set7.issubset(set8))  # Returns True because every elements in set7 is present in set8. 
print(set7 <= set8)

print(set8.issubset(set7))  # Returns False because element 10 in set8 is not there in set7.
print(set8 <= set7)
```

```python
# issuperset

set7 = {7,8,9}
set8 = {7,8,9,10}

print(set8.issuperset(set7))  # Returns True because set8 contains all the elements present in set7.
print(set8 >= set7)

print(set7.issuperset(set8))  # Returns False because set7 does not contains all the elements present in set8. no 10
print(set7 >= set8)
```

In Python, sets are mutable (doesn't mean we can change the elements but we can do add, remove, discard operations right?!), but frozensets are immutable.  
A frozenset is a frozen (immutable) set. Once created, you cannot add or remove elements.Because frozenset does not support methods such as add(), remove(), or clear().

LESSON 1: WHAT IS A DICTIONARY IN PYTHON

We have to subscript the keys (like indexing) to fetch the values. 


```python
dict1 = {1:"Navin", 2: "Kiran", 4: "Harsh"}
dict1[2]

# dict1[3]        # KeyError: 3, because the key 3 is not present. To avoid this error, we can use '.get()'. 
dict1.get(3)      # It returns nothing instead of throwing an error. If you print it, will returns 'None'
print(dict1.get(3)) 

# We can also customize in case no values found for a particular key
dict1.get(3, "Not found")  # In case, no values found, it will returns 'Not found'. 
```

```python
# You can use '.zip()' to combine keys and values in a separate list. 
keys = ['Navin', 'Kiran' , 'Harsh']
values = ['Python', 'Java' ,'JS']

combo = dict(zip(keys, values))
print(combo)
```

```python
# Adding a new key value pair in the existing dictionary:

combo['Monika'] = 'CS'
print(combo)

# Deleting a particular key-value pair
del combo['Navin']
print(combo)
```

LESSON 3: HOW TO USE DICTIONARY METHODS  

11 methods -> keys(), values(), pop(), popitem(), copy(), get(), setdefault(), clear(), fromkeys(), items(), update()

```python
users = {0:'Mario', 1:'Luigui', 2:'james'}

# pop() removing the specified key-value pair. 
users.pop(1)  
print(users)

# popitem() removes the last key-value pair in the dictionary
users.popitem()
print(users)
```

```python
sample_dict = {0: ['a', 'b'], 1: ['c','d']}

# copy() 
my_copy = sample_dict.copy()
print(my_copy)

# To verify this, you can check both dict's memory location id. 
print(id(sample_dict))
print(id(my_copy))    # Both have their own memory address. 

my_copy[0][1] = '!!!'   # This refers shallow copy
print(sample_dict)
print(my_copy)     
```

###### 
Note:

Original dict is unique and the copy of dict is unique. 

However the references inside it are not unique, they still point to the same memory address for each one of the lists which means if we change a value in any one the lists, it's going to update throughout all the dictionaries.  

If you want to prevent that behaviour, you have to look deep copy, but deep copy takes a few more resources and can slow down your code base depending on what operation you might be performing. 


```python
users = {0:'Mario', 1:'Luigui', 2:'james'}

# setdefault()
print(users.get(1))
print(users.get(9))   # Missing key
print(users.get(9, 'Not found'))  # Customized value. 
print(users)    # Here with '.get()', users won't get updated. 

print(users.setdefault(9, 'Missing!'))  
print(users)    # However, here with '.setdefault()' users got updated. This is the difference. 
```

```python
# fromkeys()  -> This method helps us to create keys from the list. 

people = ['Mario', 'Luigui', 'James']

users1 = dict.fromkeys(people)
print(users1)   # Want to put any values in common for all keys? Check the below:

users1 = dict.fromkeys(people, 'unknown')  # Here, 'unknown' going to be the common values for all the keys.
print(users1)
```
