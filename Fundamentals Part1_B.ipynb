```python
# 1) Use list comprehension to create a new list with **only alphabetic strings** (no numbers in them)

mixed_data = ["apple3", "banana", "grape5", "orange2", "guava"]

# Answer: 

words_only = [words for words in mixed_data if words.isalpha()]
print(words_only)

# Explanation:
# word.isalpha() returns True only if ALL characters are letters (no digits/ symbols)
# Filters out strings containing numbers like "apple3", "orange2"
# Essential for text processing in NLP/ data cleaning tasks. 
```

```python
# 2) Use dictionary comprehension to create a new dict with **only quarters having sales > 20,000**

sales = {"Q1": 15000, "Q2": 23000, "Q3": 18000, "Q4": 27000}

# Answer: 

high_sales = {k:v for k, v in sales.items() if v > 20000}
print(high_sales)

# Explanation: 
# sales.items() gives key-value pairs (k,v)
# if v > 20000, it filters quarters with high sales
# Creates new dict with same keys but only qualifying values, perfect for KPI filtering!
```

##### We can get the same output in FOR loop also then why WHILE loop? 
FOR loop is more like a iterator in Python. Here, we the users know the number of iterations of the loop.   
Basically we use WHILE loop in those situations when the user don't know how many times these statements would be executed. See the below example.

```python
list1 = [1,2,0,5,6]   # Generally an empty list is considered FALSE.So, list1 here is TRUE.

while list1:          # Means - "Keep executing the loop as long as list1 contains at least one item."
    print("Hi Jenny")  # See here do we users know how many times this program will get executed? No right? 
                       # If we run this alone, it will be an infinite loop. 
    list1.pop()        # Removes last element each time, 
    # and when the list1 gets empty after the five iteration,it becomes false hence while loop stopped there. 
```

##### WHILE LOOP WITH ELSE BLOCK
ELSE block would be executed when the WHILE block gets FALSE.   
BUT, ELSE block would not get executed when the WHILE loop forcefully terminated using BREAK.   
Let's see the examples below:

```python
count = 1
while count <= 5:
    print(count)
    count +=1    # Here, the loop ends naturally. 
else:
    print("In Else block")  # So, the ELSE block got executed. 

# You might have a question that why 'ELSE" block also runs here after the sucessful execution of WHILE loop?
# Note that - The else belongs to the while loop, not to an if statement. Easy rule to remember: 
# while...else: The else block runs when the loop ends naturally, but not when the loop is terminated by break. 
```

```python
count = 7   # 7 is greater than 5, so the condition is false and only ELSE statement executed. 
while count <= 5:
    print(count)
    count +=1
else:
    print("In Else block")
```

```python
count = 1
while count <= 5:
    print(count)
    count +=1
    if count == 3:
        break
else:
    print("In Else block")
print("Out from loop") # This statement is out from loop, so this can be executed. n

# Note that ELSE block would not get executed because the WHILE loop has been forcefully terminated by BREAK
# Also 3 not gets printed. BREAK statement ignores the last element always. Like here count == 3. 3 gets ignored. 
```

##### Now, let's see how it differs from FOR loop and WHILE loop. 

Note the terminology 'Sentinal Loop' which means It's a special value which is used to tell the program to terminate the loop/ to "stop" or "the input is finished."

It is **commonly used with while loops** when you don't know in advance how many times the loop should run.

Easy way to remember:  
Sentinel value = A special value + Loop stops when it appears

```python
number = int(input("Enter a value (-1 to quit): "))
while number != -1:
    print(number)
    number = int(input("Enter a value (-1 to quit): "))
else:
    print("In Else block")
```

In the above example, user know that until entering -1, the program keeps asking to input a number and the condition got satisfied once the user entered -1.  
Then the WHILE loops stop execution and the ELSE block would get executed.

##### LOGICAL OPERATORS

```python
# Logical operators includes ‘AND’, ‘OR’, ‘NOT’. 
# Here, ‘And & OR’ are binary logical operators however ‘NOT’ is an Unary logical operator. 
# And -> returns TRUE only when both conditions are TRUE otherwise it returns FALSE. 
# Or -> returns TRUE if one of the conditions is TRUE. Otherwise, it returns FALSE. 
# •	Not -> returns TRUE if the conditional expression returns FALSE and vice versa. 

# Let's see an example for AND:
x = 10
y = 20
z = 30

if x > y and y < z: # Here second part of condition is True however first part is False, hence the statement doesn't got executed. 
    print("x is the largest number")
else:
    print("x is not the largest number")
```

```python
# Let's see an example for OR:

x1 = 30
y1 = 50
z1 = 10

if x1 > y1 or x1 > z1:  # Here second part of condition is True eventhough first part is False, hence the statement got executed.  
    print("x1 is atleast larger than one number")
else:
    print("x1 is the smallest number")
```

```python
# Let's see an example for NOT:

x = 10
y = 20
z = 30

if not(x > y or x < z):  # Here both the conditions are FALSE. 'NOT' operator reverse it as TRUE, hence the statement got executed. 
    print("x is not the largest number")   # Change the OR operator as AND to see the difference. 
else:
    print("x is not the smallest number")
```

```python
# Difference between '==' vs "is" (Equality vs Identity)
# is → checks object identity: "Are these two variables pointing to the exact same object?"
# == → checks value equality: "Do these two objects contain the same values?"

list_a = [1, 2, 3]
list_b = list_a
list_c = [1, 2, 3]

# 'is' (Identity)
def check_identity(list1, list2):
    return list1 is list2

print(check_identity(list_a, list_b)) 
print(check_identity(list_a, list_c))

# '==' (Equality)
def check_equality(list1, list2):
    return list1 == list2

print(check_equality(list_a, list_b)) 
print(check_equality(list_a, list_c)) 

```

```python
# 21) replace() 
str19 = "This is our example. This is our second example"
print(str19.replace("example", "demo", 1))   
# Here, note that the word 'example'(which is mentioned twice) should be replaced by 'demo'. 
# But the third argument 1 refers to the count that means that only the first occurance should get replaced. 
# Change it to 2 and you can see it has been replaced in both the places. 

# 22) Rfind() -> rfind() gives the position of the last occurance of a specific word
str19 = "This is our example. This is our second example"
print(str19.rfind("example"))  
print(str19.rfind("Java"))
# Here, we have the word 'example' twice. So, rfind method returns the index position of the specified word which occured lastly. 
# Note in second 'print' -> word 'Java' is not in the string. Hence, rfind() returns -1. 

# 23) Rindex() 
# Similiar to rfind(). Difference is when value is not found, it doesn't returns -1, instead it returns value error - "substring not found" 
str19 = "This is our example. This is our second example"
print(str19.rindex("example"))  
# print(str19.rindex("Java"))

# 24) Rjust()  
# -? used to make a string a specific width by adding characters to the left side of the string.
str20 = "American vandal"
print(str20.rjust(25,"*"))
# Here, the len of str20 is 11. In condition, we give 25. Therfore, the remaining spaces in the string will get filled by '*' on left side. 

# 25) Rsplit()
str21 = "Football, Archary, Cricket, Squash, Hockey, Volleyball"
print(str21.rsplit(",", 4))
# Note in output that it returns a split of the string into a list, starting the splitting from the right side.
```

##### C) Strings and Bytes Built-ins: 

```python
# String (str) → human-readable text; 
# Bytes (bytes) → raw data in a **fixed, immutable form**. Bytes is useful when Python needs to work with **raw data**.
# Bytearray (bytearray) → **raw data that you can modify**

# 1) Bytes: Creates immutable bytes. Bytes is used to store data as a sequence of bytes.

text = "Hello"

c = bytes(text, "utf-8")
c1 = bytes(text, "utf-16")
c2 = bytes(text, "utf-32")
print(c)
print(c1)
print(c2)

# Remember: bytes is immutable — you cannot change individual bytes. 
# If you do any element modification using index, it will throw an error -> 'bytes' object does not support item assignment
```

```python
# 2. bytearray() — Creates changeable bytes; bytearray is similar to bytes, but it is mutable.

c3 = bytearray([65, 66, 67])  # bytearray only takes integer values not 'str' object 
print(c3)
c3[1] = 68
print(c3)
```

```python
# 3) memoryview() — Views bytes without copying them. memoryview() allows you to access the underlying data of a bytes-like object.
# memoryview() does NOT work with a normal Python string (str). 
# It works with objects that support the buffer protocol, such as bytes, bytearray, some ther binary objects. 

data = bytearray(b"Hello")
view = memoryview(data)
print(view[0])   # Output returns 72 because H is represented by byte 72 in ASCII/UTF-8.

# You can also modify the original data through the memory view.
data = bytearray(b"Hello")
data[0] = 74  # If we change the first element's bytearray from 72 ('H') to 74, the character also changed ('J') when you viewed it. 
print(data)

```

```python
# 5) Open() -> Opens a file; Very important for Data Analytics because you'll work with CSV, TXT and other files.

data1 = open("data.txt", "r")
content = data1.read()
print(content)
data1.close()

# Better approach -> Use with, which automatically closes the file:

with open("data.txt", "r") as file:
    content = file.read()
print(content)
```

##### D) Iteration Built-ins: 

```python
# iter, next, aiter, anext, enumerate, zip, reversed, sorted, filter, map, all, any, range, slice

# 1) iter() -> iter() converts an iterable like a list into an iterator.
numbers3 = [10,20,30]
my_iterator = iter(numbers3)

print(next(my_iterator))  # next() -> next() gets the next value from an iterator.
print(next(my_iterator))
print(next(my_iterator))
# print(next(my_iterator))  # It gives an error because there is no further iteration
```

```python
# 2) enumerate() -> enumerate() gives you both the index and value while looping.

for index, i in enumerate(range(2, 6)):
    print(index, i)

# You can also start from 1

names1 = ["Anu", "Baama", "Charlie"]
for index, name in enumerate(names1, start = 1):
    print(index, name)
```

```python
# 3) zip() -> zip() combines two or more iterables element by element.

name4 = ["Anu", "Radha", "Ravi"]
age2 = [23,19,20]

for name , age in zip(name4, age2):
    print(name, age)

# You can also create a dictionary:
my_dict = dict(zip(name4, age2))
print(my_dict)

# Very useful for data processing.
```

```python
# 4) reversed() -> reversed() gives the items in reverse order.
age2 = [23,19,20]

for age in reversed(age2):
    print(age)

# you can convert this in to a list. 
list2 = list(reversed(age2))
print(list2)

# Important: reversed() does not modify the original list. It copies. 
```

```python
# 5) sorted() -> sorted() sorts values from ascending to descending and returns a new list.
age2 = [23,19,20]
sorted1 = sorted(age2)
print(sorted1)

# If you need descending to ascending, use 'reverse= True' argument
sorted2 = sorted(age2, reverse=  True)
print(sorted2)

# 6) filter() -> filter() keeps only the values that satisfy a condition.

# Example: Get only the even numbers using filter function
numbers4 = [1,2,3,4,5,6,7,8,9]
even = filter(lambda x : x % 2 == 0, numbers4)
print(list(even))

# Example 2: Get all the numbers greater than 4
condition = filter(lambda x: x > 4, numbers4)
print(list(condition))
```

```python
# 7) map() -> map() applies a function to every item.

# Example: Square every number.
numbers4 = [1,2,3,4,5,6,7,8,9]
square1 = map(lambda x: x **2, numbers4)
print(list(square1))

# Another example: Get new prices by increased each value by 1.20
price = [10,20,30]
new_price = map(lambda x: x + 1.20, price)
print(list(new_price))

# Remember:  map() → transform every value -> filter() → select some values
```

```python
# 8) all() -> all() checks whether EVERY value is True.

numbers4 = [1,2,3,4,5,6,7,8,9]
even1 = all(number % 2 == 0 for number in numbers4)  # 'all' takes exactly one arguement
print(even1)                                         # all values should be TRUE

numbers5 = [2,4,6,8,10]
even2 = all(number % 2 == 0 for number in numbers5)
print(even2)

# 9) any() -> any() checks whether ATLEAST ONE value is True.
numbers5 = [1,3,5,8]
even1 = any(number % 2 == 0 for number in numbers4)  # 'any' takes exactly one arguement
print(even1)                                         # Atleast ANY of the one value must be True

numbers6 = [1,3,5,7]
odd = any(number % 2 == 0 for number in numbers6)
print(odd)
```

```python
# 10) range() -> range() generates a sequence of numbers, commonly used with loops.
# start, stop, step
for i in range(2,11,2):
    print(i)

# reverse:
for i in range(10, 0, -2):
    print(i)
```

```python
# 11) slice() -> slice() creates a slicing object that can be used to extract part of a sequence.

numbers5 = [2,4,6,8,10]
abc = numbers5[1:4]
print(abc)

# The equivalent using slice():
numbers5 = [2,4,6,8,10]
myslic = slice(1, 4)
result = numbers5[myslic]
print(result)

# You can also specify a step:
numbers5 = [2,4,6,8,10]
my_slic = slice(1,5,2)
result1 = numbers5[my_slic]
print(result1)
```

##### F) Object Built-ins: 

```python
# object, getattr, setattr, delattr, hasattr, dir, id, hash, len, isinstance, issubclass, callable, super, type.
```

```python
# 1) object -> object() creates the most basic Python object.

obj = object()
print(obj)
print(type(obj))

# 👉 object is the base class of almost all Python classes.
```

```python
# 2) getattr() -> getattr() gets an attribute/value from an object.

class Student:
    name = "Dhivya"   # Here, name is an attribute of the Student class.

student = Student()   # Here, student is an object. 
print(getattr(student, "name"))  # Instead of 'student.name', we can use: getattr(student, "name")

# We can write with a default value
print(getattr(student,"age" ,25))  # If age doesn't exist, it returns 25.

# Things to remember:
# Class = blueprint (Student)
# Object = actual thing created from the blueprint (student)
# Attribute = data/property (name)
# Method = function belonging to a class/object (def function's callable one)
```

```python
# 3) setattr() -> setattr() sets/creates an attribute.

class Student:
    pass   # Here, we don't mention any attributes, instead we placed 'pass' keyword

student = Student()   # Here, method = class
setattr(student, "name","Bharathi")  
print(student.name)

# Note that 'setattr' expected only 3 arguments.
```

```python
# delattr() -> delattr() deletes an attribute.

class Student():
    name = "Anu"

student = Student()
print(student.name)

# delattr(student, "name")  # Just put it in comments to avoid error. 
# print(student.name)       # AttributeError: 'Student' object has no attribute 'name'
                            # because name has been deleted.

# This is similar to: del student.name
```

```python
# 5) hasattr() -> hasattr() checks whether an object has a particular attribute.

class Student():
    name = "Dhivya"

student = Student()
print(hasattr(student, "name"))
print(hasattr(student, "age"))

# Very useful when you want to check before accessing an attribute.
```

```python
# 6) dir() -> dir() shows the attributes and methods available for an object.

numbers = [10, 20, 30]
print(dir(numbers))

# You'll get a long list such as in the output. 
print("remove" in dir(numbers))

# 👉 dir() is very useful while learning and debugging.
```

```python
# 7) id() -> id() returns the unique identity of an object during its lifetime.

name5 = "Bharath"
print(id(name5))

# You can compare two objects:

name5 = "Bharath"
name6 = name5
print(id(name6))

# Both IDs will be the same because b refers to the same object.
```

```python
# 8) hash() -> hash() returns a hash value for a hashable object.

name5 = "Bharath"
print(hash(name5))

# ⚠️ The exact value can change between Python runs.
# Strings, numbers, and tuples containing hashable objects can be hashed. But a list cannot

print(hash(10))
print(hash("Python"))
# print(hash([1, 2, 3]))  # TypeError: unhashable type: 'list'
```

```python
# 9) isinstance() -> isinstance() checks whether an object belongs to a particular class/type.

age3 = 19

print(isinstance(age3, int))
print(isinstance(age3, str))

# Another example:

value = 100

if isinstance(value, str):
    print("Numeric value")
else:
    print("Non Numeric value")
```

```python
# 10) issubclass() -> issubclass() checks whether one class is derived from another class.

class Animal():
    pass
class Dog(Animal):
    pass
print(issubclass(Dog, Animal))

# Another example:
print(issubclass(int, object))   # True because int ultimately inherits from object.
```

```python
# 11) callable() -> callable() checks whether something can be called using ().

def greet():
    print("Hello")

print(callable(greet))

# A number is not callable:

number2 = 10

print(callable(number2))
```

```python
# 12) super() -> super() is used to access methods/attributes from a parent class.

class Animal:
    def speak(self):   # Here, speak is a method
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        super().speak()    # calls the parent's speak() method.
        print("Dog barks")

dog = Dog()

dog.speak()
```

##### G) Descriptor Built-ins: 

```python
# classmethod, property, staticmethod. 
```

```python
# 1) classmethod() -> A classmethod is a method that receives the class itself as its first argument, usually called cls.

class Student:   
    school = "ABC school"   # Attribute

    @classmethod
    def show_school(cls):   # cls refers to the Student class.
        print(cls.school)

Student.show_school()  # calls the method directly using the class.

# Easy memory trick
# classmethod → works with CLASS
# cls         → represents CLASS
```

```python
# 2) staticmethod() -> A static method is a method that doesn't need the class or object.
# It behaves almost like a normal function, but we keep it inside the class because it is related to that class.

class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10,20))

# Notice there is no self and no cls.
# Easy memory trick: staticmethod → doesn't need object or class
```

```python
# 3) property() -> property() allows us to treat a method like an attribute.
# This sounds confusing initially, so let's see an example.

# Without property

class Student:
    def get_name(self):
        return "Dhivya"

student = Student()
print(student.get_name())  # We need parentheses because get_name is a method.

# With Property

class Student:

    @property
    def get_name(self):
        return "Dhivya"

student = Student()
print(student.get_name)  # Here, We don't use (). The method is being accessed like an attribute.


# Remember this sentence:
# classmethod → I need the CLASS.
# staticmethod → I need NEITHER class nor object.
# property → I need the OBJECT, but I want the method to LOOK LIKE AN ATTRIBUTE.
```

##### H) Dynamic / Code Built-ins:                                                                                      

```python
# eval, exec, compile, globals, locals, vars, _import_
```

```python
# 1) eval() -> Evaluate an expression. It takes a string containing a Python expr, evaluates it, & returns d result.

result2 = eval("10 + 20")
print(result2)

expression = "100 * 5 + 20"
result3 = eval(expression)
print(result3)

# This can be useful when an expression is generated dynamically.
# Important distinction:  eval() is intended for an expression, Not for a complete statement such as:

# str25 = eval("x = 10")   # This produces an error.
# print(str25)
```

```python
# 2) exec() — Execute Python code. exec() is more powerful than eval().
# It can execute Python statements or multiple lines of Python code.

code = """
x1 = 30
y1 = 10
result = x1 + y1
"""
exec(code)
print(result)

# Scenario: Dynamically generated code
# Imagine a program receives a piece of Python code as a string:

code1 = """
for i in range(3):
    print(i)
"""

exec(code1)

# Own example:

example = """
a1 = 3 * (3+2)
b1 =  4 * 2
result1 = a1 - b1
"""

exec(example)
print(result1)
```

```python
# 3) compile() —> Compiles code before executing it. It converts source code into a code object.
# (compile() means prepare Python code so that Python can execute it later)
# Think of it as: Source code → compile() → Code object → exec() / eval()
# compile() has three important modes: eval, exec and single.

#  -> "eval" : For an expression.

compiled_code = compile("10 + 20","<string>","eval") # compile(source, filename, mode)
# Here, compiled_code is not 30. It's a compiled code object.

result4 = eval(compiled_code)
print(result4)

# Why do we need compile()? -> Suppose you have Python code as a string:
code2 = "x = 10 + 20"  # Python sees this as just text.
# compile() converts that text into something Python can execute.
```

```python
# -> "exec" : For statements
compiled_code2 = compile("x = 78\nprint(x)", "<string>", "exec")
result5 = exec(compiled_code2)
print(result5)

# "single" : For a single interactive-style statement:
compiled_code3 = compile('print("Hello")', "<string>", "single") 
exec(compiled_code3)    

# Key concept to remember: "single" belongs to compile(), not to a separate single() function.

# For your current level, remember:
# eval   → expression
# exec   → statements
# single ->  interactive-style statement
# compile → prepare code for execution.  
```
