##### 120) What will be the output?

- b) Error
- c) [1,2,3,1,2,3]

Answer: C) [1,2,3,1,2,3]  
Explanation: In Python, when you multiply a list by an integer, it creates a new list that repeats the original list that many times. So, when you do a * 2, it takes the list [1, 2, 3] and repeats it twice, resulting in [1, 2, 3, 1, 2, 3].   
So, how to multiply each element of the list by 2? You can use a list comprehension to achieve this:
a = [1, 2, 3]  
result = [x * 2 for x in a]  
print(result)
