##### TASK 25: You have a list of numbers.

2) Use Pandas to create a Dataframe with these numbers and add a new column showing whether each number is above the Mean

```python
numbers7 = [10, 15, 20, 25, 30, 35]

# NumPy
import numpy as np

arr = np.array(numbers7)
print(arr)

Mean = arr.mean()
Standard_deviation = arr.std()

print("Mean: ", Mean)
print("Standard deviation: ", round(Standard_deviation, 2))

# Pandas
import pandas as pd

df = pd.DataFrame(numbers7)
print(df)

df["above_mean"] = df[0] > Mean
print(df["above_mean"])

# Explanation:
# NumPy's '.mean()' and '.std()' calculates average and spread. 
# Pandas dataframe stores data as new column and it uses here the boolean condition. 

```

```python
# Convert 10,240 seconds into hours with it's minute using 'timedelta' function. 
# Rememeber that 1 hour has 3600 seconds ( 1 hour = 60 min * 60 sec => 1 hour = 3600 seconds)

# NOTE: 'timedelta' does not have an .hours or .minutes attribute. It has .days, .seconds, and .microseconds. 
# Always use .total_seconds() to calculate.
  # To get minutes, divide total seconds by 60.
  # To get hours, divide total seconds by 3600.

from datetime import timedelta

total_seconds = 10240

td = timedelta(seconds = total_seconds)

hours = td.seconds // 3600
minutes = (td.seconds  % 3600) // 60

print("Hours: ", hours)
print("Minutes: ", minutes)

# Explanation:
# Hours → 10240 // 3600 = 2
# Leftover → 10240 % 3600 = 3040
# Minutes → 3040 // 60 = 50
# Final time: 2 hours, 50 minutes
```

```python
# Another example: Convert 5 hours into minutes with it's seconds using 'timedelta' function.

# NOTE: 'timedelta' does not have an .hours or .minutes attribute. It has .days, .seconds, and .microseconds. 
# Always use .total_seconds() to calculate.
  # To get minutes, divide total seconds by 60.
  # To get hours, divide total seconds by 3600.

from datetime import timedelta

hours = 5 

td1 = timedelta(hours = hours)

minutes = td1.total_seconds() // 60
seconds = td1.total_seconds()


print("Minutes: ", int(minutes))
print("Seconds: ", int(seconds))
```

```python
# Anoter example: 
# You have 9500 seconds. Using Python’s timedelta function, calculate and print:
# The total hours (as a float).The total minutes (integer). The remaining seconds after full minutes.

from datetime import timedelta

total_seconds = 9500

td2 = timedelta(seconds = total_seconds)

hours1 = td2.total_seconds() / 3600
minutes1 = (td2.total_seconds() % 3600) // 60
seconds1 = td2.total_seconds()

print("Hours: ", round(hours1, 2))
print("Minutes: ", int(minutes1))
print("Seconds: " , int(seconds1))

```

##### Strings

```python
# Given a string "Hello world" - Convert to uppercase, Reverse the string & Count number of vowels.

string = "Hello world"

string_upper = string.upper()
print(string_upper)

reversed_string = string[::-1]
print(reversed_string)

count_vowels = sum(1 for char in string.lower() if char in 'aeiou')
print(count_vowels)
```

```python
# Check if a string is a palindrome.

string1 = "level"

is_palindrome3 = string1 == string1[::-1]
print(is_palindrome3)

# Another way: Case-insensitive palindrome check

string2 = "Level"

if string2.lower() == string2.lower()[::-1]:
    print("Yes, It's palindrome")
else:
    print("It's not palindrome")
```

```python
# ONLY REMOVE ALL SPECIAL CHARACTERS from a string (keep only letters and numbers).

import re  # (regular expression) library.

text12  = "Hello@#!! Python123 %%&& World*()"

cleaned = re.sub(r"[^A-Za-z0-9]" ,'', text12)   # re.sub(pattern, replacement, string). Here,'sub' means substitute
print(cleaned)

# How it works:
# [^A-Za-z0-9] = match anything NOT a letter or number
# re.sub(..., '', text) = replace all those characters with empty string
```

```python
# Remove Only Punctuation for the following text. 

text13 = "Hello!!! How are you?? I'm fine... :)"

import string

removed_punc = ''.join(char for char in text13 if char not in string.punctuation)
print(removed_punc)
```

```python
# Remove Only Digits

text14 = "Room123 is on floor 4"

removed_dig = ''.join(ch for ch in text14 if not ch.isdigit())
print(removed_dig)
```

```python
# Remove Only letters

text14 = "Room123 is on floor 4"

removed_letters = ''.join(ch for ch in text14 if not ch.isalpha())
print(removed_letters)
```

```python
# Keep Spaces but Remove Special Characters

text15 = "Hello @World! Python#3 is $awesome :)"

import re
removed_spaces2 = re.sub(r"[^A-Za-z0-9 ]", "", text15)  # Note here the list bracket closed after a space. 
print(removed_spaces2)
```

```python
# Remove ALL vowels from a string

text16 = "Data Scientist"

import re

# Remove vowels (A, E, I, O, U — both upper & lower case)
removed_vowels = re.sub(r"[aeiouAEIOU]", "", text16)  # re.sub(pattern, replacement, string). Here,'sub' means substitute

# Additional practice:
alter = re.sub(r"[^aeiouAEIOU]", "", text16)   # Note the tilt symbol. 

print(removed_vowels)
print(alter)
```

```python
# Given a list of numbers, find: Max, Min, Sum, Average

data1 = [1, 2, 3, 2, 4, 1, 6]

print("Maximum: ", max(data1))
print("Minumum: ", min(data1))
print("Sum: ", sum(data1))
print("Average: ", sum(data1)/ len(data1))

# Note: list does not have 'avg' or 'mean' built-in functions to calculate average. 
```

```python
# Remove duplicates from a list (3 different ways).

data1 = [1, 2, 3, 2, 4, 1, 6]

# Method 1: Using SET (Fastest, but order not preserved)
removed_dup = set(data1)

print(removed_dup)

# Method 2: Using a loop (Order preserved ; Best choice if you want to keep original order).

unique_list = []

for item in data1:
    if item not in unique_list:
        unique_list.append(item)

print(unique_list)

# Method 3: Using dict.fromkeys() (Order preserved & clean)

unique_list1 = list(dict.fromkeys(data1))

print(unique_list1)
```

##### TASK 68: Count the frequency of each character in a string. 



```python
text = "Data Science"

from collections import Counter
freq_count2 = Counter(text)
print(freq_count2)


# Another method
freq = {}

for char in text:
    freq[char] = freq.get(char, 0) + 1  # If the character appears once that means it didn't appeared 

print(freq)

```
