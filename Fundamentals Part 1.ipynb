```python
# BIT TOUGH TO UNDERSTAND. HAVE TO REMEMBER THE CONCEPT AND CODE. 

# Check if number is prime: a number that can be divided exactly only by itself and 1

def is_prime(n):
    return n > 1 and all (n % i != 0 for i in range(2, int(n**0.5) + 1))

is_prime(23)

#Explanation:
   # n > 1 ensures only numbers greater than 1 are considered.
   # all(...) returns True only if all conditions inside are True.
   # (n % i != 0 for i in range(...)) checks divisibility by every number i from 2 to √n.
   # If none divide evenly, the number is prime.
```

#### SORTING WITH CUSTOM KEYS
##### In real world coding & interviews, you don't always sort just numbers.                                               Sometimes you need to sort names by length, tuples by the second element, or even dictionary values.                  That's where custom keys comes in. 

##### The sorted() function lets you define how you want things sorted using the key argument.                          You give it a function (or lambda) that tells Python what to sort by.

##### We can use Sorting with Custom keys when: Sorting leadership by scores,                                        Sorting logs by timestamps,                                                                                          Sorting dictionaries by values &                                                                                  Rearranging custom data formats. 

```python
# Example 1: Sort a list of tuples by the second element. 
# Each tuple has two elements. We want to sort them by the second one (x[1])

pair = [(1,3), (2,1), (4,2),(3,0)]

sorted_pair = sorted(pair, key = lambda x: x[1])
print(sorted_pair)

# See the output, the second element in each tuple is in the order like 0,1,2,3. 
```

```python
# Another example

pair2 = [(2,3,5),(1,5,3),(4,1,2),(8,9,4),(2,6,1)]

sorted_pair2 = sorted(pair2, key = lambda x: x[2])
print(sorted_pair2)

sorted_pair3 = sorted(pair2, key = lambda x: x[0])
print(sorted_pair3)
```

```python
# Example 2: Sort String by length. 

names = ["Alice", "Bob", "Charlie"]

sorted_names = sorted(names, key = len)
print(sorted_names)

sorted_names1 = sorted(names, key = len, reverse = True)
print(sorted_names1)

# We used len as the key function, So Python sorted the strings by their length. 
```

```python
# Example 3: Sort words by last character using 'def' function.  

words = ["apple", "carrot","banana"]

def sort_by_last_char(word):
    return word[-1]

print(sorted(words, key = sort_by_last_char))

# Note thoroughly by the output that the sorting based on not from last order, but based on last character of each element. 
```

```python
students = [("Alice", 85), ("Bob", 72), ("Charlie",91)]

# Write code that will sort the students by their marks in descending order. 

sorted_students = sorted(students, key = lambda x: x[1], reverse = True)
print(sorted_students)

# Another example: Sort the students by their ranks in ascending order

Students1 = [("Gayu", 9), ("Dharan", 5), ("Uma", 2)]
sorted_Students1 = sorted(Students1, key = lambda x: x[1])
print(sorted_Students1)
```

### DECORATORS (INTRO LEVEL)

#### Decorators allow you to modify the behaviour of functions without changing their actual code. They are commonly used for logging, authentication, measuring execution time, etc...

```python
# Basic syntax:

def decorator_function(original_func):
    def wrapper():
        print("Before the function runs")
        original_func()
        print("After the function runs")
    return wrapper
    
@ decorator_function
def say_hello():
    print("Hello!")

say_hello()

# What's happening here?
   # Well, the @decorator_func is shorthand for say_hello = decorator_func(say_hello). 
   # It wraps say_hello() with additinal behaviour defined in the wrapper()

# When is this useful?
   # Adding logging without changing function code. 
   # Checking permissions before running a function. 
   # Retrying a failed function etc.. 
```

```python
# Another example: 

def example_deco(org_func):
    def wrap2():
        print("Absolute")
        org_func()
        print("Certainly")
    return wrap2

@ example_deco
def say_hi():
    print("Hey")

say_hi()

# When you observe the ouput, you can see that 'Absolute' and 'Certainly' got added with 'Hey' when we call say_hi. 
```

```python
def short(func):
    def wrapper():
        return func().upper()
    return wrapper
    
@short
def say():
    return "hello"

say()
```

#### ENCAPSULATION: Hides internal details. Achieved using private variables. It refers to the binding of data and code together into a single unit. 

```python
#### Have  you ever seen a capsule used for medicion? 
#### It is made up of 2 components. First is a medicine itself and second is the outer shape which is used to protect medicine from various factors. 
#### So, important medicines are wrapped in to some container called capsule which is used to protect it from outside contact. 

#### This concept is known as Encapsulation. Similarly, in OOP - variables and functions are wrapped into class/objects. 
#### You can consider class as a outer container which protects functions and variables from any outside unauthorized access. 
#### So, wrapping a function or variable into single unit like a class is called Encapsulation in OOP. 
#### It is used to restricting access to the function and it is used to data hiding/ abstraction. 

#### Abstraction - Hiding irrelevant or confidetial information and revealing only neccessary part to the users. 
#### Example - Only audience watching cricket lively in stadium would know what's exactly happening there. 
#### If audience watch via mobile phone or televison, abstraction there would be happened that means media operatorss may cut the unimportant parts or any confidential event happening in ground and only showing the remaining content of the match. 

### _balance ~ single underscore means protection of the data
### __balance ~ double underscore means: it's privacy that indicates even more strong protection. 
```

```python
class bank_account:
    def __init__(self, balance):
        self.__balance = balance   # makes it private with __

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance or Invalid amount")


acc = bank_account(15000)
print(acc.get_balance())

acc.deposit(5000)
print(acc.get_balance())

acc.withdraw(25000)
print(acc.get_balance)
```

##### 5) Inheritance:
Child class inherits features from parent class. 

```python
# Case 1: 
# Child DOES NOT need new attributes. You do NOT need to write __init__ in the child.
# Parent __init__ is automatically used

class Animal:                              # Parent Class
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

class Dog(Animal):                         # Child Class
    def speak(self):
        return f"{self.name} always Bow Bow!"
    
d3 = Dog("Tommy", "Husky")
print(d3.breed)
print(d3.speak())
```

```python
# Case 2:
# Child NEEDS its own initialization (MOST COMMON).
# You MUST use __init__ and call the parent constructor using super().

class Employee:
    def __init__(self, name):
        self.name = name 

class DataScientist(Employee):
    def __init__(self, name, tools):    # Here, we newly added parameter called 'tools' in child class. 
        super().__init__(name)          # call parent constructor by mentioning 'super()'
        self.tools = tools       

d4 = DataScientist("Rahul", ["Python", "SQL"])  
print(d4.name)
print(d4.tools)

# When creating object (d4) by mentioning child class, we must input all the values of exising parameters & the newly added parameters) 
# Note that we can input the additional parameter in child class (tools)
# However we need to input the exisiting parameter (name) from parent class by using super()
# When we don't use super() and call the (name) parameter in child class, it returns an attribute error. 
# Example below. 
```

```python
# Don't run this. 

# Case 3:  
# Child overrides __init__ but DOES NOT call super()

class DataScientist(Employee):
    def __init__(self, tools):
        self.tools = tools       # Newly added parameter that's not in parent class. 

d5 = DataScientist("Anvar")      # Here, we call 'name' attribute which is not there in a child class. 
print(d5.name)           # Eventhough it inherits, we should call it by using 'super()' when we adding a new attribute. 

# AttributeError: 'DataScientist' object has no attribute 'name'
```

📌 Interview-ready answer

In inheritance, __init__ should be used when the child class needs its own initialization.                                  When overriding it, the parent’s constructor must be called using super() to ensure proper initialization.


🔥 Best Practice Rule (remember this)

✔ No new attributes → no need for __init__  
✔ New attributes → define __init__ + call super()  
❌ Never override without super()

```python
# Generators

def countdown(n):
    while n > 0:
        yield n
        n -= 1

for num in countdown(4):
    print(num)

# Explanation: 
# We created a def function called 'countdown' and  passing n value as parameter. 
# Now we write condition as: 'n' value must be greater than 0, this is condition is true, then it will move to next line. 
# For this 'True' condition, we used 'while' loop in it. 
# Being generator, we used 'yield' method for generating 'n'. 
# Now we write 'for' loop by passing the countdown function's value as 4. 
# As per the condition we pre-defined, values staring printed as n value that is 4. 
# Then, (4 -=1 = 3), (3 -=1 = 2).. It will continues until the 'n' number of times that is 4. 
```

```python
# Fibonacci Generator. 

def fibonacci():
    a, b = 0, 1
    for _ in range(5):
        yield a
        a, b = b, a+ b

for num in fibonacci():
    print (num)
```

##### 3) Context Managers (with):
Use **with** to manage resources like files or database connections safely and cleanly. 

```python
with open("output.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Custom Context Manager:

class OpenFile:
    def __init__(self, filename):
        self.filename = filename

    def __enter__(self):
        self.file = open(self.filename, "r")
        return self.file

    def __exit__(self, exc_type, exc_value, exc_tb):  # Runs always, even if an exception occurs. 
                                                      # exc_type, exc_value, exc_tb capture exceptions (if any)
        self.file.close()   # Closes the file safely  

with OpenFile("output.txt") as f:
    print(f.read())
```
